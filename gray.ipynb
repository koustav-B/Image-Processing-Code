```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("flower.jpg")

# Convert BGR to RGB for displaying with matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Convert to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display both
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Color Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread("flower.jpg", 0)

# Negative transformation
negative = 255 - image

# Display original and negative images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(negative, cmap="gray")
plt.title("Negative Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the grayscale image
image = cv2.imread("flower.jpg", 0)

# Log transformation
c = 255 / np.log(1 + np.max(image))
log_image = c * np.log(1 + image)

# Convert to 8-bit image
log_image = np.array(log_image, dtype=np.uint8)

# Display images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(log_image, cmap="gray")
plt.title("Log Transformed Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
image = cv2.imread("flower.jpg", 0)

# Normalize pixel values between 0 and 1
normalized = image / 255.0

# Gamma value
gamma = 2

# Gamma transformation
gamma_image = np.power(normalized, gamma)

# Convert back to 0-255
gamma_image = np.uint8(gamma_image * 255)

# Display images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gamma_image, cmap="gray")
plt.title("Gamma Transformed Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
image = cv2.imread("flower.jpg", 0)

# Define intensity range
low = 100
high = 180

# Create a copy
boosted = image.copy()

# Boost the selected intensity range
boosted[(image >= low) & (image <= high)] = 255

# Display images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(boosted, cmap="gray")
plt.title("Intensity Boosted Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
image = cv2.imread("flower.jpg", 0)

# Control points
r1 = 80
r2 = 180
s1 = 30
s2 = 220

# Create output image
output = np.zeros_like(image)

# Region 1: r <= r1
output[image <= r1] = (s1 / r1) * image[image <= r1]

# Region 2: r1 < r <= r2
output[(image > r1) & (image <= r2)] = (
    ((s2 - s1) / (r2 - r1)) *
    (image[(image > r1) & (image <= r2)] - r1) + s1
)

# Region 3: r > r2
output[image > r2] = (
    ((255 - s2) / (255 - r2)) *
    (image[image > r2] - r2) + s2
)

# Convert to uint8
output = np.uint8(output)

# Display
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(output, cmap="gray")
plt.title("Piecewise Linear Transform")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread("flower.jpg", 0)

# Apply histogram equalization
equalized_image = cv2.equalizeHist(image)

# Display original and equalized images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(equalized_image, cmap='gray')
plt.title("Histogram Equalized Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read color image
image = cv2.imread("flower.jpg")

# Convert BGR to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Display image
plt.imshow(image_rgb)
plt.title("Color Image")
plt.axis("off")
plt.show()

# Plot histograms for R, G and B
plt.hist(image_rgb[:, :, 0].ravel(), bins=256, range=[0, 256], label="Red")
plt.hist(image_rgb[:, :, 1].ravel(), bins=256, range=[0, 256], label="Green")
plt.hist(image_rgb[:, :, 2].ravel(), bins=256, range=[0, 256], label="Blue")

plt.title("RGB Histogram")
plt.xlabel("Intensity")
plt.ylabel("Number of Pixels")
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Continuous gray-level values
r = np.linspace(0, 1, 500)

# PDF
pdf = 2 * r

# CDF / Transformation function
s = r ** 2

# Plot PDF
plt.figure(figsize=(6, 4))
plt.plot(r, pdf)
plt.title("PDF of Gray Levels")
plt.xlabel("Gray Level (r)")
plt.ylabel("p(r)")
plt.grid()
plt.show()

# Plot CDF / Transformation
plt.figure(figsize=(6, 4))
plt.plot(r, s)
plt.title("CDF / Transformation Function")
plt.xlabel("Input Gray Level (r)")
plt.ylabel("Output Gray Level (s)")
plt.grid()
plt.show()
```
