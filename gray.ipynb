```python
import cv2
import matplotlib.pyplot as plt

# Read image
image = cv2.imread("flower.jpg")

# Convert BGR to RGB for displaying with matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Convert to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display both
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Original Color Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray_image, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread("flower.jpg", 0)

# Negative transformation
negative = 255 - image

# Display original and negative images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(negative, cmap="gray")
plt.title("Negative Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the grayscale image
image = cv2.imread("flower.jpg", 0)

# Log transformation
c = 255 / np.log(1 + np.max(image))
log_image = c * np.log(1 + image)

# Convert to 8-bit image
log_image = np.array(log_image, dtype=np.uint8)

# Display images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(log_image, cmap="gray")
plt.title("Log Transformed Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
image = cv2.imread("flower.jpg", 0)

# Normalize pixel values between 0 and 1
normalized = image / 255.0

# Gamma value
gamma = 2

# Gamma transformation
gamma_image = np.power(normalized, gamma)

# Convert back to 0-255
gamma_image = np.uint8(gamma_image * 255)

# Display images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gamma_image, cmap="gray")
plt.title("Gamma Transformed Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read grayscale image
image = cv2.imread("flower.jpg", 0)

# Define intensity range
low = 100
high = 180

# Create a copy
boosted = image.copy()

# Boost the selected intensity range
boosted[(image >= low) & (image <= high)] = 255

# Display images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(boosted, cmap="gray")
plt.title("Intensity Boosted Image")
plt.axis("off")

plt.show()
```
