```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("image.jpg", 0)

# Apply log transformation
c = 255 / np.log(1 + np.max(img))
log_transform = c * np.log(1 + img)

# Convert to uint8
log_transform = np.uint8(log_transform)

# Display images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(log_transform, cmap="gray")
plt.title("Log Transformed Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image in grayscale
img = cv2.imread("image.jpg", 0)

# Normalize pixel values to 0-1
normalized = img / 255.0

# Set gamma value
gamma = 0.5

# Apply power-law transformation
gamma_transform = np.power(normalized, gamma)

# Convert back to 0-255
gamma_transform = np.uint8(gamma_transform * 255)

# Display images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gamma_transform, cmap="gray")
plt.title("Gamma Transformed Image")
plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image in grayscale
img = cv2.imread("image.jpg", 0)

# Create bit planes
bit_planes = []

for i in range(8):
    bit_plane = (img >> i) & 1
    bit_plane = bit_plane * 255
    bit_planes.append(bit_plane)

# Display all bit planes
plt.figure(figsize=(12, 8))

for i in range(8):
    plt.subplot(2, 4, i + 1)
    plt.imshow(bit_planes[i], cmap="gray")
    plt.title("Bit Plane " + str(i))
    plt.axis("off")

plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("image.jpg")

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Convert to binary image
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Find contours
contours, hierarchy = cv2.findContours(
    binary,
    cv2.RETR_EXTERNAL,
    cv2.CHAIN_APPROX_SIMPLE
)

# Draw contours
output = img.copy()
cv2.drawContours(output, contours, -1, (0, 255, 0), 2)

# Convert BGR to RGB for displaying
output = cv2.cvtColor(output, cv2.COLOR_BGR2RGB)

# Display result
plt.imshow(output)
plt.title("Detected Contours")
plt.axis("off")
plt.show()

print("Number of contours:", len(contours))
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image
img = cv2.imread("image.jpg")

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Apply thresholding
_, binary = cv2.threshold(
    gray, 0, 255,
    cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
)

# Create kernel
kernel = np.ones((3, 3), np.uint8)

# Remove noise
opening = cv2.morphologyEx(
    binary,
    cv2.MORPH_OPEN,
    kernel,
    iterations=2
)

# Find sure background
sure_bg = cv2.dilate(
    opening,
    kernel,
    iterations=3
)

# Distance transform
dist_transform = cv2.distanceTransform(
    opening,
    cv2.DIST_L2,
    5
)

# Find sure foreground
_, sure_fg = cv2.threshold(
    dist_transform,
    0.5 * dist_transform.max(),
    255,
    0
)

sure_fg = np.uint8(sure_fg)

# Find unknown region
unknown = cv2.subtract(
    sure_bg,
    sure_fg
)

# Label connected components
_, markers = cv2.connectedComponents(
    sure_fg
)

# Add 1 to all labels
markers = markers + 1

# Mark unknown region as 0
markers[unknown == 255] = 0

# Apply Watershed
markers = cv2.watershed(
    img,
    markers
)

# Mark boundaries in red
img[markers == -1] = [0, 0, 255]

# Convert BGR to RGB
result = cv2.cvtColor(
    img,
    cv2.COLOR_BGR2RGB
)

# Display result
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(
    cv2.imread("image.jpg"),
    cv2.COLOR_BGR2RGB
))
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result)
plt.title("Watershed Segmentation")
plt.axis("off")

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("image.jpg")

# Convert BGR to RGB
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Reshape image into a list of pixels
pixel_values = img_rgb.reshape((-1, 3))

# Convert to float32
pixel_values = np.float32(pixel_values)

# Define stopping criteria
criteria = (
    cv2.TERM_CRITERIA_EPS + cv2.TERM_CRITERIA_MAX_ITER,
    100,
    0.2
)

# Number of clusters
K = 3

# Apply K-Means
_, labels, centers = cv2.kmeans(
    pixel_values,
    K,
    None,
    criteria,
    10,
    cv2.KMEANS_RANDOM_CENTERS
)

# Convert centers to uint8
centers = np.uint8(centers)

# Replace each pixel with its cluster center
segmented = centers[labels.flatten()]

# Reshape back to original image size
segmented = segmented.reshape(img_rgb.shape)

# Display images
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(segmented)
plt.title("K-Means Segmentation")
plt.axis("off")

plt.show()
```
